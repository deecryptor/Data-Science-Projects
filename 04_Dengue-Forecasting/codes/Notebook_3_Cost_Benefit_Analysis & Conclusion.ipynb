## 2. Conclusion

**Summary**

* The best model that is derived for NEA's consideration to be used for predicting dengue cases is a **ARIMA model** using **weekly dengue cases between 2014 and 2022 (with outlier 2020 period omitted)**. The model has a **MAPE score of 0.10**, which implies an error rate of 10% in prediction of dengue cases for a **6 months time horizon**.
  
* The model predicts a **25.52% decrease** in dengue cases for the second half of 2022 compared to the first half of 2022. This prediction can deviate by ~10% based on the model's MAPE score of 0.10, which means the drop in cases may be as high as 18% decrease or as low as 33% decrease compared to previous 6 months time frame.
  
* With this in mind, NEA can consider **reducing manpower by 15%** which translates to a cost savings of **~ $416,344.32** per 6 months. The figure is proposed to buffer for margin of error in prediction as well as to account for any unforeseen circumstances that may cause a spike in dengue cases. If the number of dengue cases were to stay constant for another 6 months and NEA continue with the reduced manpower, the projected annual cost savings would be **~ $832,688.64**.

**Limitations and Future Works**

* Based on findings through research and data exploration, it is observed that there could be multiple factors at play simultaneously that have an impact on the number of dengue cases. As such, weather data and/or search trend data alone is not sufficient to explain for the number of dengue cases. 
  
* Factors such as the number and distribution of Aedes mosquito breeding habitats, the type of dominant Dengue virus serotype, human population density of the area with high breeding habitats, lifestyle habits of the residents in that area, and the geographical characteristic of that area (for e.g. drains or uneven surfaces that may cause water to stagnate easily) are some examples that may have an impact on the number of dengue cases.
  
* For future improvement, we can consider how to gather these data and incorporate these factors into building a more robust model for predicting dengue cases.
